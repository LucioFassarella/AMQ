````python
'''
Módulos do Python & Métodos do Qiskit
'''

import numpy as np
from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister
import qiskit.quantum_info as qi
from qiskit.circuit.library import QFTGate, HamiltonianGate

'''
Métodos auxiliares

    Métodos para quantizar vetores e operadores hermitianos

    quantiza_vetor(b_vetor = list)
        -> Algoritmo que codifica um vetor b em amplitude, retornando o circuito que prepara o estado |b > normalizado

    quantiza_unitario(unitario = list)
        -> Algoritmo que quantiza um operador unitário U, retornando um circuito que implementa U

    quantiza_expoperador(operador_hermitiano = list, tempo = float)
        -> Algoritmo que quantiza um operador hermitiano A em dado tempo t, retornando o circuito que prepara o unitário exp(-2.pi.i.A.t)
'''

def quantiza_vetor(b_vetor = list):
    '''
    Algoritmo que codifica um vetor b em amplitude, retornando o circuito que prepara o estado |b > normalizado

    Entrada:
        b_vetor (list): vetor

    Saída:
        qc (QuantumCircuit): circuito quântico que prepara o estado |b > normalizado
    '''

    # Número de qubits para representar o vetor
    num_qubits = int(np.ceil(np.log2(len(b_vetor))))

    # Normalização e completamento do vetor até a coordenada 2**n
    vetor_normalizado = np.array(b_vetor/np.linalg.norm(b_vetor)) # Normaliza o vetor
    vetor_normalizado_completo = np.zeros(2**num_qubits, dtype=complex) # Vetor nulo, ensure complex dtype
    vetor_normalizado_completo[:len(vetor_normalizado)] = vetor_normalizado

    # Circuito de inicialização
    qc = QuantumCircuit(num_qubits)
    qc.initialize(vetor_normalizado_completo, normalize = True)

    return qc

def quantiza_unitario(unitario = list):
    '''
    Algoritmo que quantiza um operador unitário U,
    retornando um circuito que implementa U

    Entrada:
        U (list): operador unitário

    Saída:
        qc (QuantumCircuit): circuito quântico que implementa U
    '''

    dimensao = len(unitario)

    num_qubits = int(np.ceil(np.log2(dimensao)))

    unitario2 = np.identity(2**num_qubits, dtype = complex)

    for i in range(dimensao):
        for j in range(dimensao):
            unitario2[i][j] = unitario[i][j]

    gate = qi.Operator(unitario2)

    qc = QuantumCircuit(num_qubits)

    qc.unitary(gate, range(num_qubits))

    return qc

def quantiza_expoperador(operador_hermitiano = list, tempo = float):
    '''
    Algoritmo que quantiza um operador hermitiano A em dado tempo t,
    retornando o circuito que prepara o unitário exp(-2.pi.i.A.t)

    Entrada:
        operador_hermitiano (list): operador
        time (float)..............: tempo

    Saída:
        qc (QuantumCircuit): circuito quântico que prepara o unitário  exp(-2.pi.i.A.t)
    '''

    dimensao = len(operador_hermitiano)

    num_qubits = int(np.ceil(np.log2(dimensao)))

    hamiltoniano = np.identity(2**num_qubits, dtype = complex)

    for i in range(dimensao):
        for j in range(dimensao):
            hamiltoniano[i][j] = operador_hermitiano[i][j]

    # Cria a porta exp(-2.pi.i.A.t)
    time = 2*np.pi*tempo
    ham_gate = HamiltonianGate(hamiltoniano, time)
    qc = QuantumCircuit(num_qubits)
    qc.append(ham_gate, range(num_qubits))

    return qc
````

````python
# Parâmetros do problema

qubits_estado = 1
qc_estado = QuantumCircuit(qubits_estado) # Aqui, não usamos o método quantiza_vetor
qc_estado.x(0)

qubits_fase = 2
qc_fase = QuantumCircuit(qubits_fase)

operador = [[1,-1/3],[-1/3,1]]
tau_1 = 1.4 # Maior que o máximo valor absoluto dos autovalores de A
qc_unitario = quantiza_expoperador(operador, tempo = 1/tau_1)
````

````python
'''
ESTIMATIVA DE FASE

Input:
    qc_unitario (QuantumCircuit)..: circuito quântico que que implementa o operador unitário
    qc_autoestado (QuantumCircuit): circuito quântico que prepara o auto-estado do operador unitário
    qubits_estado (int).......: número de qubits do circuito principal (igual ao número de qubits do qc_autoestado)
    qubits_fase (int).........: número de qubits do circuito auxiliar para armazenar os bits da fase

Output:
    qc (QuantumCircuit): circuito quântico que implementa o algoritmo Estimativa de Fase.

Métodos requeridos:
    QuantumCircuit, QuantumRegister, ClassicalRegister
    qiskit.circuit.library.QFTGate
    qiskit_ibm_runtime.SamplerV2
'''

# Cria o circuito quântico
qr_estado = QuantumRegister(qubits_estado, name = 'qr_estado')
qr_fase = QuantumRegister(qubits_fase, name = 'qr_fase')
qc = QuantumCircuit(qr_estado, qr_fase)
qc.name = 'q-QCEF'

# Aplicação de Hadamard nos qubits auxiliares
for k in range(qubits_fase):
    qc.h(qr_fase[k])

qc.barrier()

# Convert qc_unitario to a gate once for efficiency
unitario_as_gate = qc_unitario.to_gate()

# Aplicação das potências controladas do operador unitário

# U^(2^0) = U:
ctrl_gate_k0 = unitario_as_gate.control(1)

# Aplicação de U^(2^k)-controlado
qc.append(ctrl_gate_k0, [qr_fase[0]] + list(qr_estado))

# U^(2^k), para k = 1,..., num_qubits_aux -1:
for k in range(1, qubits_fase):
    # Circuito que representa U^(2^k)
    qc_unitario_potencia = QuantumCircuit(qubits_estado)
    for _ in range(2**k):
        qc_unitario_potencia.append(unitario_as_gate, list(range(qubits_estado)))

    # Circuito de U^(2^k)-controlado
    ctrl_gate_potencia = qc_unitario_potencia.control(1)

    # Aplicação de U^(2^k)-controlado
    qc.append(ctrl_gate_potencia, [qr_fase[k]] + list(qr_estado))

qc.barrier()

# Aplicação da TFQ-inversa nos qubits auxiliares
qft_gate = QFTGate(num_qubits = qubits_fase)
iqft_gate = qft_gate.inverse()
qc.append(iqft_gate, list(qr_fase))

qc.barrier()
````

````python
'''
Método Crot

    Método que implementa as rotações controladas do algoritmo HHL

    Input:
        num_qubits = int
        t0_inverso = None

    Output:
        qc (QuantumCircuit): circuito quântico que implementa as rotações controladas do algoritmo HHL

'''

# Definição de t0

t0_inverso = 2.5 # Maior do que tau_1 divididos pelo valor absoluto mínimo dos autovalores de A

# Acréscimo do registro auxiliar

qr_aux = QuantumRegister(1, name = 'qr_aux')
qc.add_register(qr_aux)

# Rotações controladas


from qiskit.circuit.library import RYGate

# The loop should iterate over the possible states of the phase estimation register
for k_phase_val in range(1, 2**qubits_fase):

    # angulo
    # The binary string for the control state must have length equal to qubits_fase
    bin_string_for_ctrl = bin(k_phase_val)[2:].zfill(qubits_fase)
    bin_string_invertida = bin_string_for_ctrl[::-1]

    # 'numero' should represent the scaled eigenvalue from the phase register
    numero = k_phase_val / (2**qubits_fase)
    seno = 1/(numero*t0_inverso)
    if np.abs(seno) > 1:
        seno = np.sign(seno) # Cap at 1 or -1

    angulo = 2*np.arcsin(seno)

    # rotação controlada

    ry_gate= RYGate(angulo)
    # num_ctrl_qubits is qubits_fase, ctrl_state's length must match it
    crot = ry_gate.control(qubits_fase, ctrl_state = bin_string_invertida)

    # ação
    # The controls are qr_fase, and the target is qr_aux[0]
    qc.append(crot, list(qr_fase) + [qr_aux[0]])

qc.barrier()
````

````python
'''
Estimativa de fase inversa
'''

# Aplicação da TFQ nos qubits auxiliares
qft_gate = QFTGate(num_qubits = qubits_fase)
qc.append(qft_gate, list(qr_fase))

qc.barrier()

# Convert qc_unitario to a gate once for efficiency
unitario_dagger_as_gate = qc_unitario.inverse().to_gate()

# Aplicação das potências controladas do operador unitário

# U^(2^0) = U:
ctrl_gate_k0 = unitario_dagger_as_gate.control(1)

# Aplicação de U^(2^k)-controlado
qc.append(ctrl_gate_k0, [qr_fase[0]] + list(qr_estado))

# U^(2^k), para k = 1,..., num_qubits_aux -1:
for k in range(1, qubits_fase):
    # Circuito que representa U^(2^k)
    qc_unitario_potencia = QuantumCircuit(qubits_estado)
    for _ in range(2**k):
        qc_unitario_potencia.append(unitario_as_gate, list(range(qubits_estado)))

    # Circuito de U^(2^k)-controlado
    ctrl_gate_potencia = qc_unitario_potencia.control(1)

    # Aplicação de U^(2^k)-controlado
    qc.append(ctrl_gate_potencia, [qr_fase[k]] + list(qr_estado))

qc.barrier()

# Aplicação de Hadamard nos qubits auxiliares
for k in range(qubits_fase):
    qc.h(qr_fase[k])
````

````python
num_qubits = qubits_estado + qubits_fase + 1

estadovetor_dict = {bin(k)[2:].zfill(num_qubits): estadovetor[bin(k)[2:]] for k in range(2**num_qubits)}

estadovetor_posselecionado_1_dict = {bin(k)[2:].zfill(num_qubits): estadovetor[bin(k)[2:]] for k in range(2**(num_qubits-1),2**num_qubits)}

norma2 = 0
for key in estadovetor_posselecionado_1_dict:
    norma2 += np.abs(estadovetor_posselecionado_1_dict[key])**2

norma = np.sqrt(norma2)

print(f"Norma = {norma}")

estadovetor_posselecionado_1_normalizado_dict = {key : estadovetor_posselecionado_1_dict[key]/norma for key in estadovetor_posselecionado_1_dict.keys()}

estadovetor_posselecionado_1_normalizado_abs_dict = {key : np.abs(estadovetor_posselecionado_1_normalizado_dict[key]) for key in estadovetor_posselecionado_1_normalizado_dict.keys()}

display(plot_histogram(estadovetor_posselecionado_1_normalizado_abs_dict))

print("Verificação da norma:")
norma_12 = 0
for key in estadovetor_posselecionado_1_normalizado_dict.keys():
    norma_12 += np.abs(estadovetor_posselecionado_1_normalizado_dict[key])**2

norma_1 = np.sqrt(norma_12)
print(f"Norma = {norma_1}")
````
